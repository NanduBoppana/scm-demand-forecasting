```python
import pandas as pd, matplotlib.pyplot as plt

df = pd.read_csv("../data/train.csv", parse_dates=["date"])
print(df.shape)  # expect (913000, 4)

s = (df[(df.store == 1) & (df.item == 1)]
       .sort_values("date").set_index("date")["sales"])

print(len(s), s.isna().sum())                      # expect 1826, 0
print(s.index.to_series().diff().value_counts())   # expect all 1 day

s.to_csv("../data/series.csv", header=True)
ax = s.plot(figsize=(12,4), title="Store 1, Item 1: daily sales")
plt.savefig("../outputs/01_full_series.png", dpi=150)
```
