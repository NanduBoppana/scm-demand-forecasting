```python
import pandas as pd, matplotlib.pyplot as plt

s = pd.read_csv("../data/series.csv", parse_dates=["date"], index_col="date")["sales"]

fig, ax = plt.subplots(3, 1, figsize=(12, 11))
s.resample("ME").mean().plot(ax=ax[0], title="Monthly mean (trend + yearly seasonality)")
s.groupby(s.index.dayofweek).mean().plot.bar(ax=ax[1], title="Average by day of week (0=Mon)")
s.groupby(s.index.month).mean().plot.bar(ax=ax[2], title="Average by month")
plt.tight_layout()
plt.savefig("../outputs/02_eda.png", dpi=150)

print(s.describe())
print("Mean by year:\n", s.groupby(s.index.year).mean())
```
